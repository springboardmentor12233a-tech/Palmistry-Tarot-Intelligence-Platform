```python
import random

deck = cards.copy()

print("Deck size:", len(deck))
```

1. Creating the Virtual Tarot Deck

Explanation:

The tarot dataset contains information about all 78 tarot cards. In this step, a virtual tarot deck is created by making a copy of the loaded card list. This allows the system to perform card shuffling and random selection while keeping the original dataset unchanged.

The virtual deck acts like a real tarot deck where cards can be randomly drawn during a reading.

```python
drawn_card = random.choice(deck)

print("Card:", drawn_card["name"])
print("Keywords:", drawn_card["keywords"])
```

2. Random Card Selection

A tarot reading begins by randomly selecting cards from the virtual deck. The random selection process simulates the process of drawing physical tarot cards.

The selected card contains complete information from the JSON dataset, including its name, keywords, meanings, and other associated details.

```python
orientation = random.choice(["Upright", "Reversed"])
print("Orientation:", orientation)
if orientation == "Upright":
  meaning = drawn_card["meanings"]["light"]
else:
  meaning = drawn_card["meanings"]["shadow"]
  print("\nMeaning:")
for m in meaning: print("-", m)
```

Card Orientation Assignment (Upright/Reversed)

Each selected tarot card is randomly assigned an orientation: Upright or Reversed.

The dataset provides two types of interpretations:

Light meaning → used for upright cards
Shadow meaning → used for reversed cards

A reversed card does not always represent something negative; it can indicate challenges, delays, internal conflicts, or areas requiring attention.
Retrieving Card Meaning from Dataset

Explanation:

Based on the assigned card orientation, the system retrieves the corresponding interpretation from the tarot JSON dataset.

For an upright card, the system extracts the light meanings, while for a reversed card, it extracts the shadow meanings.

This ensures that every generated reading is based on the original tarot knowledge stored in the dataset.

```python
# Taking user details for the tarot reading

name = input("Enter your name: ")

question = input("Enter your question: ")

category = input("Enter reading category (Career/Love/General): ")


print("\nUser Details")
print("Name:", name)
print("Question:", question)
print("Category:", category)
```

5. User Input and Reading Category

Explanation cell for PPT:

The tarot system begins by collecting user information and their query. The user's name and question provide context for generating a personalized reading. A reading category is also selected to understand the user's area of interest, such as career, relationships, or general guidance.

The collected information will later be combined with the selected tarot cards to generate a contextual interpretation.

```python
# Selecting the tarot spread type

spread = input(
    "Choose spread (single/ppf): "
)


if spread.lower() == "single":

    number_of_cards = 1


elif spread.lower() == "ppf":

    number_of_cards = 3


else:

    print("Invalid spread selected")
    number_of_cards = 1


print("Number of cards to draw:", number_of_cards)
```

6. Tarot Spread Selection

Explanation cell for PPT:

A tarot spread defines how many cards are drawn and how they are interpreted.

The prototype supports two spreads:

Single Card Spread
Draws one card
Provides general guidance related to the user's question
Past-Present-Future Spread
Draws three cards
Each card represents:
Past influences
Present situation
Future guidance

```python
# Creating a fresh copy of the deck before drawing cards
available_deck = deck.copy()

# Randomly selecting required number of cards
selected_cards = random.sample(
    available_deck,
    number_of_cards
)

print("Selected Cards:")

for card in selected_cards:
    print("-", card["name"])
```

7. Drawing Tarot Cards According to Spread
Explanation cell for PPT:

The system randomly selects tarot cards from the virtual deck based on the user's chosen tarot spread.

For a Single Card Spread, one card is drawn to provide general guidance.
For a Past-Present-Future Spread, three different cards are drawn and assigned to different life stages:
Past → Previous influences and experiences
Present → Current situation and challenges
Future → Possible direction and guidance

The random selection simulates the process of drawing cards from a physical tarot deck.

```python
# Assigning positions to selected cards

if spread.lower() == "ppf":

    positions = [
        "Past",
        "Present",
        "Future"
    ]

else:

    positions = [
        "Guidance"
    ]


# Combining cards with their positions

reading_cards = []

for i, card in enumerate(selected_cards):

    reading_cards.append({
        "position": positions[i],
        "card": card
    })


# Display result

for item in reading_cards:

    print(
        item["position"],
        "→",
        item["card"]["name"]
    )
```

8. Assigning Spread Positions
Explanation cell for PPT:

After selecting cards, each card is assigned a position based on the selected tarot spread.

For a Past-Present-Future spread:

First card represents Past
Second card represents Present
Third card represents Future

This helps organize the interpretation and create a meaningful reading structure.

```python
print("~"*60)
print("TAROT READING")
print("~"*60)
print("Name:", name)
print("Question:", question)
print("Category:", category)
if spread.lower() == "ppf":
  print("Spread: Past-Present-Future")
else:
  print("Spread: Single Card")
for item in reading_cards:

    print("\n" + "-"*50)

    card = item["card"]

    # Assigning orientation at display time
    orientation = random.choice(["Upright", "Reversed"])

    print("Position:", item["position"])
    print("Card:", card["name"])
    print("Orientation:", orientation.lower())

    print("Keywords:", ", ".join(card["keywords"]))


    if orientation == "Upright":
        meaning = card["meanings"]["light"]
    else:
        meaning = card["meanings"]["shadow"]


    print("\nDataset meanings:")

    for m in meaning:
        print("-", m)
```

8. Generate Structured Tarot Reading Output

The selected tarot cards are converted into a structured reading format containing user details, question, category, spread type, card positions, orientation, keywords, and dataset-based meanings.

This structured output improves readability and prepares the tarot interpretation data for further processing by an external LLM.

```python
from google.colab import userdata
from groq import Groq

GROQ_API_KEY = userdata.get("GROQ_API_KEY")

client = Groq(api_key=GROQ_API_KEY)

print("Groq API key loaded successfully.")
```

Loading the Groq API Key

The Groq API key is securely stored using Google Colab Secrets instead of being written directly in the notebook. During execution, the key is retrieved from the secure storage and used to authenticate requests to the external Large Language Model (LLM). This approach protects sensitive credentials while allowing secure API access.

```python
prompt = f"""
You are a friendly and thoughtful tarot reader.

The reading is for self-reflection and entertainment only.
Do not make guaranteed predictions or claim certainty.
Provide an encouraging and balanced interpretation.

User Name: {name}
Question: {question}
Category: {category}
Spread: {spread}

Selected Tarot Cards:

"""

for item in reading_cards:

    card = item["card"]

    # Assign orientation
    orientation = random.choice(["Upright", "Reversed"])

    # Select meaning
    if orientation == "Upright":
        meaning = card["meanings"]["light"]
    else:
        meaning = card["meanings"]["shadow"]


    prompt += f"""
Position: {item['position']}
Card: {card['name']}
Orientation: {orientation}
Keywords: {", ".join(card['keywords'])}

Meanings:
"""

    for m in meaning:
        prompt += f"- {m}\n"


prompt += """

Using only the information above:

Generate a tarot reading with these sections:

1. Overall Interpretation:
Explain how all cards connect together and relate to the user's question.

2. Card-wise Explanation:
Explain each card separately according to its position.
Mention:
- What the card represents
- How its orientation affects the meaning
- How it relates to the user's situation

3. Guidance:
Provide balanced self-reflection points.

4. Reflection Question:
End with one thoughtful question for the user.

Remember:
- This is for self-reflection and entertainment.
- Do not make guaranteed predictions.
- Keep the tone supportive and conversational.

"""


print(prompt)
```

Preparing the Prompt for the External LLM

The structured tarot reading is converted into a text prompt before being sent to the external LLM. The prompt includes the user's question, selected spread, card positions, orientations, keywords, and dataset meanings. The LLM uses this information to generate a personalized and conversational tarot interpretation while preserving the original meanings from the tarot dataset.

```python
models = client.models.list()

for model in models.data:
    print(model.id)
```

```python
response = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
        {
            "role": "user",
            "content": prompt
        }
    ],
    temperature=0.7,
    max_tokens=1000
)

llm_reading = response.choices[0].message.content
conversation_context = prompt + "\n\nPrevious AI Reading:\n" + llm_reading

print(llm_reading)
```

Generating Conversational Tarot Interpretation Using LLM

The structured tarot prompt is sent to an external Large Language Model through the Groq API. The LLM analyzes the user's question along with the selected tarot cards and their meanings, then generates a natural and personalized interpretation.

The LLM does not generate new tarot meanings; it uses the dataset-provided meanings and focuses on converting them into a coherent conversational response.

```python
follow_up = input(
    "\nAsk a question about your reading (or type exit): "
)
```

Interactive Follow-up Reading

The system allows users to ask additional questions after receiving the initial tarot interpretation. The follow-up questions are answered using the same selected tarot cards and previous reading context, maintaining consistency throughout the conversation.

```python
while follow_up.lower() != "exit":

    follow_up_prompt = f"""

You are a thoughtful tarot assistant.

Previous tarot reading context:

{conversation_context}


User follow-up question:

{follow_up}


Answer the user's question using the same tarot cards.
Do not draw new cards.
Do not make guaranteed predictions.
Keep the response supportive and reflective.

"""


    response = client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=[
            {
                "role": "user",
                "content": follow_up_prompt
            }
        ],
        temperature=0.7,
        max_tokens=700
    )


    answer = response.choices[0].message.content

    print("\nAI Response:")
    print(answer)


    follow_up = input(
        "\nAsk another question (or type exit): "
    )
```
