```python
import random

def get_random_tarot_card(cards):

    card = random.choice(cards)

    return {
        "name": card.get("name", ""),
        "number": card.get("number", ""),
        "arcana": card.get("arcana", ""),
        "keywords": card.get("keywords", []),
        "fortune": card.get("fortune_telling", []),
        "upright": card.get("meanings", {}).get("light", []),
        "reversed": card.get("meanings", {}).get("shadow", []),
        "affirmation": card.get("Affirmation", "No affirmation available")
    }
```

```python
reading = get_random_tarot_card(cards)

reading
```

```python
import json

with open("selected_tarot_card.json", "w") as file:
    json.dump(reading, file, indent=4)

print("Selected tarot card saved successfully!")
```

```python
def generate_tarot_interpretation(card):

    interpretation = f"""
============================================================
                 TAROT CARD INTERPRETATION
============================================================

Card Name:
{card['name']}

Card Number:
{card['number']}

Arcana:
{card['arcana']}

Suit:
{card['suit']}

Keywords:
{', '.join(card['keywords'])}

------------------------------------------------------------
Positive Meanings
------------------------------------------------------------
{chr(10).join('• ' + meaning for meaning in card['meanings']['light'])}

------------------------------------------------------------
Challenges / Shadow Meanings
------------------------------------------------------------
{chr(10).join('• ' + meaning for meaning in card['meanings']['shadow'])}

------------------------------------------------------------
Fortune Telling
------------------------------------------------------------
{chr(10).join('• ' + item for item in card['fortune_telling'])}

------------------------------------------------------------
Affirmation
------------------------------------------------------------
{card['Affirmation']}
------------------------------------------------------------
Questions for Reflection
------------------------------------------------------------
{chr(10).join('• ' + question for question in card['Questions to Ask'])}
"""

    return interpretation
```
