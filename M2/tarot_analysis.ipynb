```python
import json
import random

with open(r"C:\Users\souri\Downloads\Palmistry & Tarot Intelligence\Datasets\Data Files\tarot-images.json") as f:   # <-- edit path if needed
    tarot_data = json.load(f)

deck = tarot_data["cards"]
deck_by_name = {c["name"]: c for c in deck}

print(f"Loaded {len(deck)} cards.")
```

```python
def get_card_interpretation(card, upright=True):
    """Build readable interpretation text for a single card."""
    orientation = "Upright" if upright else "Reversed"
    meaning_list = card["meanings"]["light"] if upright else card["meanings"]["shadow"]
    meaning_text = "; ".join(meaning_list[:4])   # first 4 meanings - keeps it readable
    keyword_text = ", ".join(card["keywords"][:4])

    return {
        "card_name": card["name"],
        "orientation": orientation,
        "keywords": keyword_text,
        "meaning": meaning_text
    }

# quick test on one known card
test_result = get_card_interpretation(deck_by_name["The Fool"], upright=False)
test_result
```

```python
def draw_cards(n, seed=None):
    """Randomly draw n unique cards, each randomly upright or reversed."""
    rng = random.Random(seed)
    drawn_names = rng.sample(list(deck_by_name.keys()), n)
    drawn = []
    for name in drawn_names:
        upright = rng.choice([True, False])
        drawn.append((deck_by_name[name], upright))
    return drawn

# draw 3 cards to test
sample_draw = draw_cards(3, seed=1)
[(c["name"], "Upright" if u else "Reversed") for c, u in sample_draw]
```

```python
def generate_reading(drawn_cards, positions=None):
    """Generate the full formatted reading text."""
    lines = []
    for i, (card, upright) in enumerate(drawn_cards):
        interp = get_card_interpretation(card, upright)
        position_label = positions[i] if positions else f"Card {i+1}"
        lines.append(f"<{position_label}: {interp['card_name']} ({interp['orientation']})>")
        lines.append(f"Keywords: {interp['keywords']}")
        lines.append(f"Meaning: {interp['meaning']}")
        lines.append("")
    return "\n".join(lines)
```

```python
single = draw_cards(1)
print(generate_reading(single))
```

```python
three = draw_cards(3)
print(generate_reading(three, positions=["Past", "Present", "Future"]))
```
