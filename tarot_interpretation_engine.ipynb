The tarot reading module allows users to choose between a Single Card Spread and a Past–Present–Future Spread. When the Single Card option is selected, the system randomly draws one card to generate a concise reading. For the Past–Present–Future option, three distinct cards are selected, where each card is assigned a predefined role: Past represents previous events and influences, Present describes the current circumstances, and Future provides potential outcomes and recommendations.

```python
import random

# Introduction for the LLM
prompt = f"""
You are a compassionate tarot companion who helps users reflect on their thoughts and experiences.

Interpret the selected cards in a positive and encouraging manner.
Focus on offering guidance for self-reflection rather than making
certain predictions about the future.
Keep the conversation warm, respectful, and uplifting.
User Details
------------
Name      : {user["Name"]}
Question  : {user["Question"]}
Category  : {user["Category"]}
Spread    : {spread}

Selected Tarot Cards
--------------------
"""

# Store orientation mapping
meaning_type = {
    "Upright": "light",
    "Reversed": "shadow"
}

# Add each selected card to the prompt
for reading in reading_cards:

    card = reading["card"]

    orientation = random.choice(list(meaning_type.keys()))

    meanings = card["meanings"][meaning_type[orientation]]

    prompt += (
        f"\nPosition    : {reading['position']}\n"
        f"Card        : {card['name']}\n"
        f"Orientation : {orientation}\n"
        f"Keywords    : {', '.join(card['keywords'])}\n"
        f"Meanings:\n"
    )

    prompt += "\n".join(f"• {text}" for text in meanings)
    prompt += "\n"

# Final instructions for the LLM
prompt += """

Generate the reading using only the information provided above.

Required Sections

Using the provided tarot card information, generate a meaningful reading with the following structure:

1. Reading Summary
Provide an overall interpretation by connecting all the selected cards and explaining how they relate to the user's question.

2. Individual Card Insights
Describe each card separately by including:
• The significance of the card.
• The influence of its orientation (Upright/Reversed).
• Its relevance to the user's current situation.

3. Personal Guidance
Offer constructive suggestions that encourage self-awareness, reflection, and positive decision-making.

4. Self-Reflection Prompt
Conclude with a thoughtful question that encourages the user to reflect on their circumstances.

Guidelines:
• Base the interpretation only on the supplied card information.
• Keep the reading supportive and balanced.
• Avoid making definite predictions or promises about future events.
• Present the response in a friendly and conversational manner.
"""

print(prompt)
```

```python
response = client.chat.completions.create(
    model="openai/gpt-oss-120b",
    messages=[
        {
            "role": "user",
            "content": prompt
        }
    ],
    temperature=0.7,
    max_tokens=1000
)

llm_reading = response.choices[0].message.content
conversation_context = prompt + "\n\nPrevious AI Reading:\n" + llm_reading

print(llm_reading)
```

```python
import os
from reportlab.lib.pagesizes import A4
from reportlab.lib.styles import getSampleStyleSheet
from reportlab.lib.enums import TA_CENTER
from reportlab.lib.units import inch
from reportlab.platypus import (
    SimpleDocTemplate,
    Paragraph,
    Spacer,
    Image
)

# -----------------------------
# Styles
# -----------------------------
styles = getSampleStyleSheet()

title_style = styles["Title"]
title_style.alignment = TA_CENTER

heading_style = styles["Heading2"]
text_style = styles["BodyText"]


# -----------------------------
# Helper Functions
# -----------------------------

def add_title(story):
    story.append(Paragraph("AI Tarot Reading Report", title_style))
    story.append(
        Paragraph(
            "Generated using AI-powered Tarot Card Interpretation",
            text_style
        )
    )
    story.append(Spacer(1, 20))


def add_reading_details(story, name, question, category, spread):

    story.append(Paragraph("1. Reading Details", heading_style))

    details = [
        ("Name", name),
        ("Question", question),
        ("Category", category),
        ("Spread", spread)
    ]

    for key, value in details:
        story.append(
            Paragraph(
                f"<b>{key}:</b> {value}",
                text_style
            )
        )

    story.append(Spacer(1, 20))


def add_cards(story, reading_cards, cards_folder):

    story.append(Paragraph("2. Cards Drawn", heading_style))
    story.append(Spacer(1, 10))

    for card_data in reading_cards:

        card = card_data["card"]

        image_path = os.path.join(
            cards_folder,
            card["img"]
        )

        if os.path.exists(image_path):

            story.append(
                Image(
                    image_path,
                    width=2.3 * inch,
                    height=3.8 * inch
                )
            )

        story.append(
            Paragraph(
                f"<b>Position:</b> {card_data['position']}",
                text_style
            )
        )

        story.append(
            Paragraph(
                f"<b>Card:</b> {card['name']}",
                text_style
            )
        )

        story.append(
            Paragraph(
                f"<b>Orientation:</b> {card_data['orientation']}",
                text_style
            )
        )

        story.append(
            Paragraph(
                f"<b>Keywords:</b> {', '.join(card['keywords'])}",
                text_style
            )
        )

        story.append(
            Paragraph(
                "<b>Meaning</b>",
                heading_style
            )
        )

        for meaning in card_data["meaning"]:
            story.append(
                Paragraph(
                    "• " + meaning,
                    text_style
                )
            )

        story.append(Spacer(1, 20))


def add_ai_interpretation(story, interpretation):

    story.append(
        Paragraph(
            "3. AI Interpretation",
            heading_style
        )
    )

    interpretation = interpretation.replace("**", "").replace("*", "")

    for line in interpretation.split("\n"):

        line = line.strip()

        if line:

            if line.endswith(":"):

                story.append(
                    Paragraph(
                        line,
                        heading_style
                    )
                )

            else:

                story.append(
                    Paragraph(
                        line,
                        text_style
                    )
                )

    story.append(Spacer(1, 20))


def add_follow_up(story, history):

    if not history:
        return

    story.append(
        Paragraph(
            "4. Follow-up Conversation",
            heading_style
        )
    )

    story.append(Spacer(1, 10))

    for chat in history:

        story.append(
            Paragraph(
                f"<b>User:</b> {chat['question']}",
                text_style
            )
        )

        story.append(
            Paragraph(
                f"<b>AI:</b> {chat['answer']}",
                text_style
            )
        )

        story.append(Spacer(1, 10))


def add_disclaimer(story):

    story.append(
        Paragraph(
            "Disclaimer",
            heading_style
        )
    )

    story.append(
        Paragraph(
            "This AI-generated tarot reading is intended only for self-reflection "
            "and entertainment purposes. It should not be considered professional "
            "advice regarding health, finance, legal matters, or relationships.",
            text_style
        )
    )


# -----------------------------
# Main Function
# -----------------------------

def generate_tarot_pdf(
        pdf_path,
        name,
        question,
        category,
        spread,
        reading_cards,
        ai_interpretation,
        follow_up_history=None
):

    if follow_up_history is None:
        follow_up_history = []

    cards_folder = "/content/dataset/cards"

    doc = SimpleDocTemplate(
        pdf_path,
        pagesize=A4
    )

    story = []

    add_title(story)

    add_reading_details(
        story,
        name,
        question,
        category,
        spread
    )

    add_cards(
        story,
        reading_cards,
        cards_folder
    )

    add_ai_interpretation(
        story,
        ai_interpretation
    )

    add_follow_up(
        story,
        follow_up_history
    )

    add_disclaimer(story)

    doc.build(story)

    print("Tarot Report Generated Successfully!")

    return pdf_path
```
