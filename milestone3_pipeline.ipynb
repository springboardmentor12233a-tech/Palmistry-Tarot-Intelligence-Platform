```python
"""
Milestone 3 Pipeline - Palmistry & Tarot Intelligence Platform

Full flow: Milestone 2's combined reading -> Milestone 3's AI analysis ->
personalized dashboard PDF. This is the single entry point that ties
Milestone 2 and Milestone 3 together.

Usage (fill in your real paths/key, then run this file directly):
    python milestone3_pipeline.py
"""

import reading_report
import ai_interpretation_engine
import dashboard_report


def run_full_pipeline(
    palm_image_path,
    tarot_deck_path,
    palmistry_code_dir,
    tarot_question=None,
    spread_size=3,
    gemini_api_key=None,
    tarot_model="gemini-3.6-flash",
    analysis_model="gemini-3.6-flash",
    dashboard_output_path="dashboard_report.pdf"
):
    """Runs Milestone 2 + Milestone 3 in one call, returns the dashboard PDF path."""

    # --- Milestone 2: combined palm + tarot reading ---
    print("Step 1/3: Generating palm + tarot reading (Milestone 2)...")
    combined_reading = reading_report.generate_combined_reading(
        palm_image_filename=palm_image_path,
        tarot_deck_path=tarot_deck_path,
        palmistry_code_dir=palmistry_code_dir,
        tarot_question=tarot_question,
        spread_size=spread_size,
        use_llm=(gemini_api_key is not None),
        gemini_api_key=gemini_api_key
    )
    print(f"  Palm success: {combined_reading['palm_success']}")
    print(f"  Tarot source: {combined_reading['tarot_source']}")

    # --- Milestone 3: AI interpretation, personality, recommendations, life trends ---
    print("Step 2/3: Running AI interpretation engine (Milestone 3)...")
    analysis = ai_interpretation_engine.generate_full_analysis(
        combined_reading, api_key=gemini_api_key, model=analysis_model
    )
    print(f"  Analysis source: {analysis['_source']}")

    # --- Personalized dashboard ---
    print("Step 3/3: Building personalized dashboard PDF...")
    dashboard_path = dashboard_report.save_dashboard_report(
        analysis, combined_reading=combined_reading, output_path=dashboard_output_path, user_question=tarot_question
    )
    print(f"  Dashboard saved to: {dashboard_path}")

    return combined_reading, analysis, dashboard_path


if __name__ == "__main__":
    # ---- EDIT THESE ----
    PALM_IMAGE_PATH = r"C:\Users\souri\Downloads\Palmistry & Tarot Intelligence\palmistry-main\palmistry-main\code\input\Hand_0001657.jpg"
    TAROT_DECK_PATH = r"C:\Users\souri\Downloads\Palmistry & Tarot Intelligence\Datasets\Data Files\tarot-images.json"
    PALMISTRY_CODE_DIR = r"C:\Users\souri\Downloads\Palmistry & Tarot Intelligence\palmistry-main\palmistry-main\code"
    GEMINI_API_KEY = 'AIzaSyA913pd6G_Bof0O-H41XraqjYr2txRsR4Y'  # paste your key here to enable live LLM calls; leave None to test the fallback path
    # ---------------------

    run_full_pipeline(
        palm_image_path=PALM_IMAGE_PATH,
        tarot_deck_path=TAROT_DECK_PATH,
        palmistry_code_dir=PALMISTRY_CODE_DIR,
        tarot_question="What should I focus on this month?",
        gemini_api_key=GEMINI_API_KEY
    )
```
