```python
import json

import pandas as pd
from openai import OpenAI
```

```python
palm_lines = pd.read_csv("Palm_Line_Features.csv")
palm_features = pd.read_csv("palm_feature_interpretation.csv")
tarot = pd.read_csv("tarot_interaction.csv")

print("Files Loaded Successfully")
```

```python
palm_summary = palm_lines.iloc[0].to_dict()
feature_summary = palm_features.iloc[0].to_dict()
tarot_summary = tarot.iloc[0].to_dict()
print(feature_summary)
```

```python
client = OpenAI(
    api_key="your_openrouter_api_key_here",
    base_url="https://openrouter.ai/api/v1"
)
from getpass import getpass

from openai import OpenAI

api_key = getpass("Enter your OpenRouter API Key: ")
```

```python
prompt = f"""
You are an AI Palmistry and Tarot Interpretation Assistant.

You have been given structured outputs from a palm analysis engine.

Palm Line Features:
{json.dumps(palm_summary, indent=2)}

Palm Feature Interpretation:
{json.dumps(feature_summary, indent=2)}

Tarot Reading:
{json.dumps(tarot_summary, indent=2)}

Based on these inputs, generate a personalized report.

Return ONLY valid JSON in the following format:

{{
  "personality": "...",
  "career_guidance": "...",
  "relationship_insights": "...",
  "health_wellness": "...",
  "life_trend": "...",
  "strengths": [
      "...",
      "..."
  ],
  "areas_for_improvement": [
      "...",
      "..."
  ],
  "recommendations": [
      "...",
      "..."
  ]
}}

Important:
- Do not claim supernatural certainty.
- Present the reading as guidance for self-reflection and entertainment.
- Return ONLY JSON.
"""
```

```python
try:
    response = client.chat.completions.create(
        model="inclusionai/ling-3.0-flash:free",
        messages=[
            {
                "role": "system",
                "content": "You are an AI Palmistry Assistant."
            },
            {
                "role": "user",
                "content": prompt
            }
        ]
    )

    print(response)

except Exception as e:
    print(e)
```

```python
result = response.choices[0].message.content

print(result)
```

```python
report = json.loads(result)

print("="*70)
print("🌟 AI PALMISTRY & TAROT REPORT 🌟")
print("="*70)

print("\n🧠 Personality")
print(report["personality"])

print("\n💼 Career Guidance")
print(report["career_guidance"])

print("\n❤️ Relationship Insights")
print(report["relationship_insights"])

print("\n🏥 Health & Wellness")
print(report["health_wellness"])

print("\n📈 Life Trend")
print(report["life_trend"])

print("\n⭐ Strengths")
for item in report["strengths"]:
    print("✔", item)

print("\n⚠ Areas for Improvement")
for item in report["areas_for_improvement"]:
    print("•", item)

print("\n🎯 Recommendations")
for item in report["recommendations"]:
    print("👉", item)
```

```python
import json

with open("Final_AI_Report.json","w") as f:
    json.dump(report, f, indent=4)

print("Report Saved Successfully")
```
