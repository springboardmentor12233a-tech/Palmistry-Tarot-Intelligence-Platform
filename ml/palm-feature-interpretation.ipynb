```python
import pandas as pd
```

```python
df = pd.read_csv("/kaggle/input/datasets/alaramyaseshasri/palm-features-clean/palm_features_clean.csv")

print("Shape:", df.shape)
df.head()
```

```python
#Define Interpretation Functions
# Palm Width
def interpret_palm_width(x):
    if x < 0.25:
        return "Narrow Palm"
    elif x < 0.27:
        return "Medium Palm"
    else:
        return "Wide Palm"


# Palm Height
def interpret_palm_height(x):
    if x < 0.40:
        return "Short Palm"
    elif x < 0.45:
        return "Medium Palm Height"
    else:
        return "Long Palm"


# Finger Length
def interpret_finger(length):
    if length < 0.43:
        return "Short"
    elif length < 0.50:
        return "Medium"
    else:
        return "Long"


# Palm Shape
def interpret_aspect_ratio(r):
    if r < 1.50:
        return "Square Palm"
    elif r < 1.70:
        return "Rectangular Palm"
    else:
        return "Long Rectangular Palm"
```

```python
#Apply Interpretation
df["Palm_Width_Type"] = df["palm_width"].apply(interpret_palm_width)

df["Palm_Height_Type"] = df["palm_height"].apply(interpret_palm_height)

df["Thumb_Type"] = df["thumb_length"].apply(interpret_finger)

df["Index_Type"] = df["index_length"].apply(interpret_finger)

df["Middle_Type"] = df["middle_length"].apply(interpret_finger)

df["Ring_Type"] = df["ring_length"].apply(interpret_finger)

df["Little_Type"] = df["little_length"].apply(interpret_finger)

df["Palm_Shape"] = df["aspect_ratio"].apply(interpret_aspect_ratio)
```

```python
df[
    [
        "filename",
        "Palm_Width_Type",
        "Palm_Height_Type",
        "Thumb_Type",
        "Index_Type",
        "Middle_Type",
        "Ring_Type",
        "Little_Type",
        "Palm_Shape",
    ]
].head(10)
```

```python
#Generate Palm Analysis Report
def generate_palm_report(row):

    report = f"""
Palm Analysis Report
--------------------
Palm Shape      : {row['Palm_Shape']}
Palm Width      : {row['Palm_Width_Type']}
Palm Height     : {row['Palm_Height_Type']}

Finger Analysis
---------------
Thumb           : {row['Thumb_Type']}
Index Finger    : {row['Index_Type']}
Middle Finger   : {row['Middle_Type']}
Ring Finger     : {row['Ring_Type']}
Little Finger   : {row['Little_Type']}
"""

    return report


df["Palm_Report"] = df.apply(generate_palm_report, axis=1)
```

```python
#Display Sample Reports
for i in range(5):
    print(df.loc[i, "Palm_Report"])
    print("=" * 80)
```

```python
#Save the Final Dataset
df.to_csv("/kaggle/working/palm_feature_interpretation.csv", index=False)

print("Palm Feature Interpretation completed successfully!")
```
